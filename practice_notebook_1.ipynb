### Python can do chained comparisons

We'll do a lot of "double conditional" checking in the first few lessons.  This section teaches you that `a<b<c` does what you think it does, but to use it with caution.

See [this reference](https://www.tutorialspoint.com/chaining-comparison-operators-in-python) on chaining comparisons.

````python
-5 < -3 < -1
````

So far, the comparisons are sensible.  But let's see what happens if we do arithmetic in a chain.  

The code below compares -1 to 6, and 6 to 6.  Check it out!

````python
-1 < 1 + 5 == 6
````

  [The "plus" operator happens before the comparisons](https://docs.python.org/3/reference/expressions.html#operator-precedence).  So first, 1+5 turns into 6.  Then, left-to-right, -1 is compared to 6 using "less than" (true).  Then, 6 is compared to 6 using "is equal to" (true).  Since both are true, the value is `True`.

I wanted to give you this little tutorial, since I could totally see someone interpreting this as the following:

````python
(-1 < 1) + (5 == 6)
````

See how the result is completely different!?!  The two comparisons `<` and `==` happen first, producing true/false values (0, 1).  Then, they're added together.  And since `True` converts to `1`, and `False` converts to `0`, the result is the number `1`.  

Neat!

An f-string (formatted string literal) is a string literal that may contain curly braces indicating replacement fields. 🧠 Predict what the code below will output, then run it to check your thinking. When using f-string formatting, curly braces represent a spot for data.  Each `{}` should have a piece of data in it.

````python
a = "apples"
b = "bananas"

message = f'{a} {b}' # curlies make the contents of a and b show up
print(message)
````

In the preceding cell, we have two variables, `a` and `b`.  When we use them in an f-string, like the one to produce `message`, they go in curlies.  Compare with the following:

````python
a = "apples"
b = "bananas"

message = f'a b' # no curlies, get the literals a and b
print(message)
````

We can also format numbers using f-string formatting.  Here we print $\pi$ to 8 decimal places:

````python
import math
pi = math.acos(-1)
pi_as_string = f'{pi:0.8f}'
print( pi_as_string )
````

### A bit on functions

Here's a simple function:

````python
def number_map(n):
    
    my_map = {1:'one', 2:'two', 3:'three', 4:'four'}
    
    return my_map[n]    
````

The function's name is `number_map`.  The name of the parameter inside the function is `n`.  Inside, the function has an internal variable called `my_map`.  It returns the string that corresponds to the input position number.

Let's try calling it!

````python
print(number_map(2))
````

The string "two" should have printed.

#### The name of a parameter doesn't have to correspond to the name of an argument

Let's define a few numbers `a`, `b`, and then pass them to the function.  🧠  Observe how none of the variables are called `n`, and yet inside the function `n` takes on the value passed into the function.

````python
a, b = 2, 3

print(f'returned value for `a`: {number_map(a)}')
print(f'returned value for `b`: {number_map(b)}')
````

Python even doesn't care if you name your variable outside the function `n`.  Inside, `n` is whatever value was passed into the function.

````python
n = 4
print(f'returned value for `n`: {number_map(n)}')
````

#### Variables defined inside a function aren't visible outside the function

Remember the variable inside the function, `my_map`.  It was defined *inside* `number_map`.  This means we can't see it, get its values, or interact with it in any way.  Trust me, this is a good thing.  More on that in Lesson 3.

````python
print(my_map) # <--- generates an error.  my_map was defined in a scope.
````
